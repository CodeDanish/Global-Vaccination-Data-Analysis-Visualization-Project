```python
# Clean up the footer row if it exists
vaccine_schedule_df = vaccine_schedule_df[vaccine_schedule_df['ISO_3_CODE'].str.len() == 3]
```

```python
"""Reason: Whenever TARGETPOP is missing, the corresponding TARGETPOP_DESCRIPTION is almost exclusively "General/routine".
           This means the vaccine is part of a standard national childhood schedule rather than being targeted at a specific subgroup.

   Solution: Impute with a logical categorical placeholder like 'GENERAL'."""
vaccine_schedule_df["TARGETPOP"].fillna("GENERAL",inplace=True)


"""Reason: This typically occurs for adult or seasonal vaccines (such as INFLUENZA_ADULT), where shots are given during annual
           campaigns or risk-group allocations rather than at a fixed age milestone (like M2 for 2 months or Y4 for 4 years).

   Solution: Impute with 'NOT_SPECIFIED' or 'VARIABLE' so it remains a discrete categorical feature for your model."""
vaccine_schedule_df["AGEADMINISTERED"].fillna("NOT_SPECIFIED",inplace=True)


"""Reason: This is an optional, qualitative free-text field used by country representatives to add specific annotations
           (e.g., "police, firemen..." or "older persons"). Free-text fields are inherently sparse.

   Solution: Impute with an empty string '' or 'NONE'. This prevents downstream natural language processing text vectorizers
             (like Word2Vec, TF-IDF, or Tokenizers) from crashing on NaN float values."""
vaccine_schedule_df["SOURCECOMMENT"].fillna("",inplace=True)
```

#**Step 2 : SQL Database Setup**

#**General Guidelines :**

● **Create Tables:** Store the extracted and cleaned data into relational SQL tables (e.g., vaccination data, disease incidence data, antigen data).

● **Normalize Data:** Structure the data into separate tables (e.g., vaccines, diseases, countries, years) to avoid redundancy and improve querying performance.

● **Data Integrity:** Implement primary and foreign keys to ensure referential integrity.

### **What is a Star Schema?**

A **Star Schema** is an optimal way to organize data for analytical reporting and dashboard building, such as in Power BI. It structures data into a central **Fact Table** surrounded by multiple **Dimension Tables**. When visualized, this layout branches out from the center, closely resembling the shape of a star.


### Understanding the Split: Fact Tables vs. Dimension Tables

To build an efficient data model, we split our columns into two distinct categories:

#### 1. Fact Tables (The Actions/Metrics)

Fact tables store the quantitative, measurable data resulting from real-world events or observations. They consist primarily of numbers (metrics) and foreign key IDs that point to the dimension tables.

* **What they answer:** "How much?" or "How many?"
* **Examples in your project:** `fact_vaccine_coverage` (storing `TARGET_NUMBER`, `DOSES`, and `COVERAGE` percentages) , `fact_reported_cases` (storing raw disease `CASES` counts) , and `fact_incidence_rates` (storing the calculated `INCIDENCE_RATE`).



#### 2. Dimension Tables (The Context/Attributes)

Dimension tables provide the descriptive context or attributes surrounding those metrics. They contain textual descriptors that you use to filter, slice, and group your dashboard visuals.

* **What they answer:** "Who?", "What?", "Where?", or "When?"
* **Examples in your project:** `dim_countries` (storing `country_name` and `who_region`) , `dim_diseases` (storing `disease_description`) , and `dim_antigens` (storing vaccine details).

```python
reported_clean = pd.read_csv("Reported_Cases_Cleaned_Data.csv")
reported_fact = reported_cases_clean[["CODE","YEAR","DISEASE","CASES"]].drop_duplicates()
print(reported_fact)
```

# **Creating SQL Script to push data in dim_incidence_rates Table**

```python
incidence_clean = pd.read_csv("Incidence_Rate_Cleaned_Data.csv")
incidence_fact = incidence_clean[["CODE","YEAR","DISEASE","DENOMINATOR","INCIDENCE_RATE"]].drop_duplicates()
print(incidence_fact)
```

```python
# Clean the codes column to ensure matching spaces don't break the join
incidence_fact["CODE"] = incidence_fact["CODE"].astype(str).str.strip()

# FILTER OUT ORPHANED ROWS: Keep only rows where the country code exists in your master list
incidence_fact_filtered = incidence_fact[incidence_fact["CODE"].isin(valid_country_codes)]
```

```python
# Open a text file to write standard SQL code
with open("insert_incidence_rates.sql", "w") as f:
    f.write("USE global_health_db;\n")

    for _, row in incidence_fact_filtered.iterrows():
        # Map your exact requested columns: code, year, disease, denominator, incidence_rate
        country_code = str(row['CODE']).replace("'", "''")
        year = int(row['YEAR'])
        disease_code = str(row['DISEASE']).replace("'", "''").strip()

        # Escape string formatting for denominator (e.g., "per 100,000 population")
        denom_val = row['DENOMINATOR']
        if pd.isna(denom_val) or denom_val is None:
            denominator = "NULL"
        else:
            denominator = f"'{str(denom_val).replace("'", "''").strip()}'"

        # Safe decimal handling for incidence_rate to preserve unit integrity
        inc_val = row['INCIDENCE_RATE']
        if pd.isna(inc_val) or inc_val is None:
            incidence_rate = "NULL"  # Safely preserves missing data as database NULL
        else:
            incidence_rate = str(float(inc_val)) # Retains decimal precision

        # Build the exact INSERT query matching your fact table schema
        f.write(f"INSERT INTO fact_incidence_rates (country_code, year, disease_code, denominator_scale, incidence_rate) "
                f"VALUES ('{country_code}', {year}, '{disease_code}', {denominator}, {incidence_rate});\n")

print("SQL script successfully created!")
```

# **Creating SQL Script to push data in fact_vaccine_coverage Table**

```python
coverage_clean = pd.read_csv("Coverage_Cleaned_Data.csv")
coverage_fact = coverage_clean[["CODE","YEAR","ANTIGEN","COVERAGE_CATEGORY","TARGET_NUMBER","DOSES","COVERAGE"]].drop_duplicates()
print(coverage_fact)
```

```python
# Clean the codes column to ensure matching spaces don't break the join
coverage_fact["CODE"] = coverage_fact["CODE"].astype(str).str.strip()

# FILTER OUT ORPHANED ROWS: Keep only rows where the country code exists in your master list
coverage_fact_filtered = coverage_fact[coverage_fact["CODE"].isin(valid_country_codes)]
```

```python
# Open a text file to write standard SQL code
with open("insert_vaccine_coverage.sql", "w") as f:
    f.write("USE global_health_db;\n")

    for _, row in coverage_fact.iterrows():
        # Map your exact requested columns: CODE, YEAR, ANTIGEN, COVERAGE_CATEGORY, TARGET_NUMBER, DOSES, COVERAGE
        country_code = str(row['CODE']).replace("'", "''")
        year = int(row['YEAR'])
        antigen_code = str(row['ANTIGEN']).replace("'", "''").strip()

        # Safely handle coverage category text
        cov_cat_val = row['COVERAGE_CATEGORY']
        if pd.isna(cov_cat_val) or cov_cat_val is None:
            coverage_category = "NULL"
        else:
            coverage_category = f"'{str(cov_cat_val).replace("'", "''").strip()}'"

        # Handle target_number (numeric integer metric)
        target_val = row['TARGET_NUMBER']
        if pd.isna(target_val) or target_val is None:
            target_number = "NULL"
        else:
            target_number = str(int(target_val))

        # Handle doses (numeric integer metric)
        doses_val = row['DOSES']
        if pd.isna(doses_val) or doses_val is None:
            doses = "NULL"
        else:
            doses = str(int(doses_val))

        # Handle coverage percentage (numeric decimal metric)
        coverage_pct_val = row['COVERAGE']
        if pd.isna(coverage_pct_val) or coverage_pct_val is None:
            coverage = "NULL"
        else:
            coverage = str(float(coverage_pct_val))

        # Build the exact INSERT query matching your fact table schema
        f.write(f"INSERT INTO fact_vaccine_coverage (country_code, year, antigen_code, coverage_category, target_number, doses_administered, coverage_percentage) "
                f"VALUES ('{country_code}', {year}, '{antigen_code}', {coverage_category}, {target_number}, {doses}, {coverage});\n")

print("SQL script successfully created! Download 'insert_vaccine_coverage.sql' from your sidebar.")
```

###2. Bar Chart: Regional Disparities (Low-Coverage Areas)

This script aggregates the average vaccination coverage across different WHO regions to highlight which geographical zones are falling behind.

```python
# Loading the data
df_vaccine_covg = coverage_fact_filtered
df_countries = country_dim

# Join vaccine data with country details to get regional grouping
df_regional = pd.merge(df_vaccine_covg, df_countries, left_on='CODE', right_on='ISO_3_CODE', how='left')

# Calculate average coverage by region
df_region_summary = df_regional.groupby('WHO_REGION')['COVERAGE'].mean().reset_index().sort_values(by='COVERAGE')

# Plot Bar Chart
plt.figure(figsize=(10, 6))
sns.barplot(data=df_region_summary, x='WHO_REGION', y='COVERAGE', palette='coolwarm')
plt.axhline(y=90, color='r', linestyle='--', label='WHO Target (90%)')
plt.title('Average Vaccination Coverage by WHO Region')
plt.xlabel('WHO Region')
plt.ylabel('Average Coverage (%)')
plt.ylim(0, 100)
plt.legend()
plt.tight_layout()
plt.show()
```

###3. Line Graph: Multi-Year Trends (Impact Assessment)

This script displays chronological trends over multiple years, illustrating how disease incidence rates dynamically respond to changes in vaccination coverage over time.

```python
# Loading the data
df_vaccine_covg = coverage_fact_filtered
df_incidence = incidence_fact_filtered

# Aggregate global averages per year
yearly_vaccine = df_vaccine_covg.groupby('YEAR')['COVERAGE'].mean().reset_index()
yearly_incidence = df_incidence.groupby('YEAR')['INCIDENCE_RATE'].mean().reset_index()
df_trends = pd.merge(yearly_vaccine, yearly_incidence, on='YEAR')

# Plot Line Graph with dual axes
fig, ax1 = plt.subplots(figsize=(12, 6))

# Primary Axis: Coverage Bars/Lines
color = 'tab:blue'
ax1.set_xlabel('Year')
ax1.set_ylabel('Average Vaccine Coverage (%)', color=color)
sns.lineplot(data=df_trends, x='YEAR', y='COVERAGE', ax=ax1, color=color, linewidth=2.5, marker='o')
ax1.tick_params(axis='y', labelcolor=color)

# Secondary Axis: Incidence Rate Trend
ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Average Disease Incidence Rate (per 100k)', color=color)
sns.lineplot(data=df_trends, x='YEAR', y='INCIDENCE_RATE', ax=ax2, color=color, linewidth=2.5, marker='s')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Global Trends: Impact of Vaccination Coverage on Disease Incidence')
fig.tight_layout()
plt.show()
```
